```python
import sys
import numpy as np
import pandas as pd
import scipy
import matplotlib.pyplot as plt
import sklearn

assert sys.version_info[:2] == (3, 12)
print('Python :', sys.version.split()[0])
print('Interpréteur :', sys.executable)
print('NumPy :', np.__version__)
print('pandas :', pd.__version__)
print('SciPy :', scipy.__version__)
print('scikit-learn :', sklearn.__version__)
```

```python
x = np.linspace(0, 2 * np.pi, 100)
table = pd.DataFrame({'x': x, 'sin(x)': np.sin(x)})
display(table.head())
plt.plot(table['x'], table['sin(x)'])
plt.title('NumPy + pandas + Matplotlib fonctionnent')
plt.xlabel('x (radians)')
plt.ylabel('sin(x)')
plt.grid(True)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression

X = np.array([[0], [1], [2], [3]])
y = np.array([1, 3, 5, 7])
model = LinearRegression().fit(X, y)
prediction = model.predict([[4]])[0]
assert np.isclose(prediction, 9)
print(f'Prédiction pour x=4 : {prediction:.1f} (attendu : 9)')
```
